# Scikit-learn

```
from sklearn.datasets import load_iris
iris=load_iris()
X=iris.data
y=iris.target
feature_names=iris.feature_names
target_names=iris.target_names
print('Feature names:',feature_names)
print('Target names:',target_names)
print('Type of X:',type(X))
print('First 5 rows of X:\n',X[:5])

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.4,random_state=1)
print('X_train shape:',X_train.shape)
print('X_test shape:',X_test.shape)
print('y_train shape:',y_train.shape)
print('y_test shape:',y_test.shape)

from sklearn.linear_model import LogisticRegression
logreg=LogisticRegression()
logreg.fit(X_train,y_train)
y_pred=logreg.predict(X_test)
from sklearn import metrics
print('Accuracy:',metrics.accuracy_score(y_test,y_pred))
```

# Tensorflow & Keras

```
import tensorflow as tf
from tensorflow.keras import layers,models
(x_train,y_train),(x_test,y_test)=tf.keras.datasets.mnist.load_data()
x_train=x_train/255.0
x_test=x_test/255.0
model=models.Sequential([
    layers.Flatten(input_shape=(28,28)),
    layers.Dense(128,activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(10,activation='softmax')
])
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
model.fit(x_train,y_train,epochs=5)
test_loss,test_accuracy=model.evaluate(x_test,y_test)
print('Test accuracy:',test_accuracy)
```

```
import numpy as np
from scipy.sparse.csgraph import dijkstra
from scipy.sparse import csr_matrix

arr = np.array([
  [0, 1, 2],
  [1, 0, 0],
  [2, 0, 0]
])

newarr = csr_matrix(arr)

print(dijkstra(newarr, return_predecessors=True, indices=0))
```

# Hugging face Transformers

```
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline

# --- 1. Tokenizer & Model Generation ---
tokenizer = AutoTokenizer.from_pretrained("t5-small")  # Load tokenizer
model = AutoModelForSeq2SeqLM.from_pretrained("t5-small")

inputs = tokenizer("summarize: Hello world", return_tensors="pt")
outputs = model.generate(**inputs, max_length=10)        # Generate tokens
print(tokenizer.decode(outputs[0], skip_special_tokens=True))  # Decode text

model.save_pretrained("./my_model")  # Save model locally

# --- 2. High-level Pipeline ---
pipe = pipeline("sentiment-analysis")  # Quick pipeline
print(pipe("Great job!"))
```

```
from fastapi import FastAPI, HTTPException, Depends, BackgroundTasks
from fastapi.testclient import TestClient
from pydantic import BaseModel

# 1. Define FastAPI App
app = FastAPI()

class InputData(BaseModel):
    value: int

def verify_token(token: str = "secret"):
    if token != "secret":
        raise HTTPException(status_code=401, detail="Invalid Token")

@app.post("/predict")
def predict(data: InputData, bg: BackgroundTasks, _=Depends(verify_token)):
    bg.add_task(print, f"Processing {data.value}")
    return {"result": data.value * 2}

# 2. Create TestClient (Replaces uvicorn + requests for Colab)
client = TestClient(app)

# 3. Make Requests directly in Colab!
response = client.post(
    "/predict",
    json={"value": 10},
    headers={"token": "secret"}
)

print("Status Code:", response.status_code)
print("Response JSON:", response.json())
```

```
import uvicorn
import threading
import requests
import nest_asyncio
from fastapi import FastAPI
from pydantic import BaseModel

nest_asyncio.apply()

# 1. Define FastAPI App
app = FastAPI()

class InputData(BaseModel):
    value: int

@app.post("/predict")
def predict(data: InputData):
    return {"result": data.value * 2}

# 2. Function to start uvicorn in background
def run_server():
    uvicorn.run(app, host="127.0.0.1", port=8000, log_level="error")

# Start server thread
server_thread = threading.Thread(target=run_server, daemon=True)
server_thread.start()

# 3. Now you can use the actual 'requests' library to talk to the local server!
res = requests.post("http://127.0.0.1:8000/predict", json={"value": 10})
print("Requests Output:", res.json())
```
